#### Rule-based / Template Generator Demo

**Concept:**
- Early "AI" systems used explicit IF–THEN rules to generate text or make decisions.
- No learning from data; all logic is hand-crafted.

**Example:** Simple expert system for weather advice.

#### More Real-life Rule-based / Template Generator Examples

Below are three practical IF–THEN text generation demos:

1. **Customer Support Auto-Reply**: Responds to user messages based on detected keywords.
2. **Simple Medical Triage**: Gives basic advice based on symptoms.
3. **Restaurant Recommendation**: Suggests a meal based on the time of day.


```python
# Customer Support Auto-Reply (Keyword-based)
def auto_reply(message):
    message = message.lower()
    if 'refund' in message:
        return "We're sorry to hear you want a refund. Please provide your order number."
    elif 'late' in message or 'delay' in message:
        return "We apologize for the delay. Your order is on its way!"
    elif 'cancel' in message:
        return "Your cancellation request has been received. We'll process it shortly."
    else:
        return "Thank you for contacting support. How can we help you?"

# Test cases
for msg in [
    "I want a refund for my order",
    "Why is my package late?",
    "Can I cancel my subscription?",
    "Hello, I have a question"
]:
    print(f"User: {msg}\nBot:  {auto_reply(msg)}\n")
```

```python
# Simple Medical Triage (Symptom-based)
def triage(symptom):
    symptom = symptom.lower()
    if 'fever' in symptom and 'cough' in symptom:
        return "You may have the flu. Please rest and consult a doctor if symptoms persist."
    elif 'headache' in symptom and 'nausea' in symptom:
        return "Possible migraine. Stay hydrated and rest."
    elif 'chest pain' in symptom:
        return "Chest pain can be serious. Seek medical attention immediately."
    else:
        return "For non-urgent symptoms, monitor your health and consult a professional if needed."

# Test cases
for s in [
    "I have a fever and cough",
    "Severe headache and nausea",
    "Mild chest pain",
    "Sore throat"
]:
    print(f"Symptom: {s}\nAdvice:  {triage(s)}\n")
```

```python
# Restaurant Recommendation (Time-based)
def recommend_meal(hour):
    if 6 <= hour < 11:
        return "How about breakfast at Sunrise Cafe?"
    elif 11 <= hour < 15:
        return "It's lunchtime! Try the Lunch Box Bistro."
    elif 15 <= hour < 18:
        return "Snack time! Visit the Coffee Corner."
    elif 18 <= hour < 22:
        return "Dinner is served at The Evening Grill."
    else:
        return "Most restaurants are closed. Try a 24/7 diner."

# Test cases
for h in [8, 12, 16, 19, 2]:
    print(f"Hour: {h}:00 -> {recommend_meal(h)}")
```

```python
# Simple rule-based text generator (expert system style)
def weather_advice(weather, temperature):
    if weather == 'rainy':
        return "It's rainy. Take an umbrella!"
    elif weather == 'sunny' and temperature > 25:
        return "It's sunny and hot. Wear sunscreen!"
    elif weather == 'sunny':
        return "It's sunny. Enjoy your day!"
    elif weather == 'snowy':
        return "It's snowy. Wear warm clothes!"
    else:
        return "Weather unknown. Be prepared!"
```

```python
# Try different scenarios
scenarios = [
    ('rainy', 18),
    ('sunny', 30),
    ('sunny', 20),
    ('snowy', -2),
    ('foggy', 10)
]

for weather, temp in scenarios:
    print(f"Weather: {weather}, Temp: {temp}°C -> {weather_advice(weather, temp)}")
```

#### N-gram (Unigram / Bigram / Trigram) Model Demo

**Concept:**
- N-gram models use frequency counts of word sequences to predict the next word.
- Unigram: single word, Bigram: two-word sequence, Trigram: three-word sequence.

**Example:** Build N-gram models from a small corpus and predict next words.

```python
from collections import Counter, defaultdict
import random
```

```python
# Small text corpus
tokens = "the cat sat on the mat the cat lay on the rug the dog sat on the mat".split()
```

```python
# Unigram model
unigram_counts = Counter(tokens)
unigram_counts
```

```python
# Bigram model
bigrams = [(tokens[i], tokens[i+1]) for i in range(len(tokens)-1)]
bigram_counts = Counter(bigrams)
bigram_counts
```

```python
# Trigram model
trigrams = [(tokens[i], tokens[i+1], tokens[i+2]) for i in range(len(tokens)-2)]
trigram_counts = Counter(trigrams)
trigram_counts
```

```python
# Predict next word using bigram model
def predict_next_bigram(word):
    candidates = [w2 for (w1, w2), count in bigram_counts.items() if w1 == word]
    
    if not candidates:
        return None
    
    return random.choice(candidates)
```

```python
print("\nBigram prediction after 'the':", predict_next_bigram('the'))
```

#### Real-World Use of Rule-based and Statistical Models

- **Expert Systems in Banking (1980s-1990s):**
  - *American Express* used rule-based expert systems for credit card fraud detection and transaction approval.
  - *Citibank* and other financial institutions used rule-based systems for loan approval and risk assessment.

- **Customer Support & IVR (Ongoing):**
  - *AT&T*, *Verizon*, and many telecoms have used rule-based Interactive Voice Response (IVR) systems for decades to route calls and answer common queries.
  - Many banks and airlines still use rule-based IVR for basic customer service.

- **Medical Diagnosis (Historical & Some Current):**
  - *MYCIN* (Stanford, 1970s-80s) was a famous rule-based system for infectious disease diagnosis.
  - Some clinical decision support tools today still use rule-based logic for alerts and reminders.

- **Spam Filtering (Early 2000s):**
  - *Yahoo! Mail* and *AOL* used rule-based and statistical (e.g., n-gram, Bayesian) models for spam detection before deep learning.

- **Search Engines (Pre-2010):**
  - *AltaVista*, *Yahoo!*, and early *Google* search used statistical models (n-gram, Markov) for query suggestions and ranking.

- **Modern Use:**
  - Many chatbots, FAQ bots, and compliance systems still use rule-based logic for reliability and transparency, especially in regulated industries.

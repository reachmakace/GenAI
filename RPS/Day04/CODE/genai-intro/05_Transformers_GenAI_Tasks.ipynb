##### How Does a Transformer Generate Text?
1. **Start with a prompt:** The user provides an initial text (e.g., "In the future, artificial intelligence will").
2. **Tokenization:** The prompt is split into tokens (words or subwords) that the model understands.
3. **Model predicts next token:** The transformer looks at the prompt and predicts the most likely next token based on its training.
4. **Append the new token:** The predicted token is added to the prompt.
5. **Repeat:** Steps 3-4 are repeated, each time using the updated text, until the desired length is reached or a stop token is produced.
6. **Detokenization:** The sequence of tokens is converted back into readable text.

This process allows the model to generate coherent and contextually relevant text, one token at a time, by always considering the full context of what has been generated so far.

##### Key Concepts Explained: Context Window, Fine-Tuning, Safety/Moderation, Multimodal

**1. Context Window**
- *Meaning:* The maximum number of tokens (words/subwords) the model can consider at once.
- *Utility:* Determines how much prior text the model can "see" when generating or understanding.
- *Example:* If the context window is 1024 tokens, the model can only use the last 1024 tokens of a conversation or document. For long documents, earlier content is ignored.
- *Need:* Longer context windows enable better understanding of long documents, code files, or conversations.

**2. Fine-Tuning**
- *Meaning:* Adapting a pre-trained model to a specific task or dataset by further training.
- *Utility:* Improves performance on domain-specific tasks (e.g., legal, medical, or company-specific language).
- *Example:* Fine-tuning GPT-2 on Shakespeare's works makes it generate Shakespearean text. Fine-tuning on customer support data creates a specialized chatbot.
- *Need:* Out-of-the-box models are generic; fine-tuning makes them expert in a target domain.

**3. Safety and Moderation**
- *Meaning:* Techniques to prevent harmful, biased, or inappropriate outputs from the model.
- *Utility:* Protects users and organizations from toxic, unsafe, or non-compliant content.
- *Example:* OpenAI's API blocks hate speech, personal data leaks, or unsafe instructions. GPT-2 (open source) has no built-in moderation.
- *Need:* Essential for public-facing applications, education, and compliance.

**4. Multimodal**
- *Meaning:* Models that can process and generate more than one type of data (e.g., text, images, audio, code).
- *Utility:* Enables richer AI applications that combine modalities (e.g., describing images, generating images from text, answering questions about pictures).
- *Example:* GPT-4V can "see" images and answer questions about them. DALL-E generates images from text prompts. CLIP matches images and captions.
- *Need:* Many real-world tasks involve multiple data types, not just text. Multimodal models unlock new GenAI capabilities.

#### How Does a Transformer Summarize Text? (Layman's Explanation)

Imagine you have a long article and you want a quick summary. A transformer model works like a super-smart assistant that:

1. **Reads the whole article carefully** – It pays attention to every sentence and word, not just the beginning or end.
2. **Finds the most important points** – It figures out what the main ideas are, just like you would highlight key sentences with a marker.
3. **Rewrites in its own words** – Instead of just copying sentences, it can rephrase and combine ideas to make the summary shorter and clearer.
4. **Checks for meaning** – It makes sure the summary still makes sense and covers the main message of the original text.

**Why is this useful?**
- Saves you time by giving you the gist of long documents.
- Helps you quickly decide if you want to read the full article.
- Works for news, research, emails, and more.

Transformers are so good at this because they can "see" the whole text at once and understand how all the parts connect, just like a human who is great at summarizing stories!

#### How Does a Transformer Know What’s Important?

A transformer model learns what’s important in text by:

1. **Learning from millions of examples:** During training, it sees many articles and their summaries, so it learns patterns about what humans consider important.
2. **Self-attention mechanism:** This lets the model "look at" all words and sentences at once, and decide which ones are most relevant to the main idea.
3. **Contextual understanding:** It understands how words and sentences relate to each other, so it can spot key facts, main arguments, and repeated themes.
4. **Scoring and weighting:** The model gives higher "attention" scores to important words and sentences, and uses these to build the summary.

**Example:**
- If an article is about a new scientific discovery, the model will pay more attention to sentences describing the discovery, results, and impact, rather than background or unrelated details.

**In short:**
- The transformer acts like a smart reader who highlights the most relevant parts, based on what it has learned from lots of real summaries and its ability to connect ideas across the whole text.

**Yes - the magic!**

When a transformer summarizes text, it uses its self-attention mechanism to calculate attention scores for every word and sentence in the input. These scores tell the model which parts of the text are most important at each step of generating the summary.

- **During summarization:**
  - The model looks at all the words at once.
  - It assigns higher attention to words and sentences that are more relevant to the main idea.
  - These attention scores help the model decide what information to keep, combine, or leave out in the summary.

**In simple terms:**
- The model is constantly "paying attention" to the most important parts of the text as it writes the summary, just like a human would focus on the key points when retelling a story.

## Question Answering (QnA) with Transformers


### Relevance and Need for QnA


- **Information Overload:** In today's world, vast amounts of information are stored in documents, websites, and databases. Manually searching for answers is time-consuming.
- **Instant Answers:** QnA systems allow users to ask natural language questions and get direct, relevant answers instantly.
- **Business Value:** Used in customer support, search engines, chatbots, knowledge management, and more, improving efficiency and user experience.
- **Accessibility:** Makes complex or lengthy information accessible to everyone, regardless of expertise.


### Popular Transformer Models for QnA


- **BERT:** The first major transformer for QnA, excels at extracting answers from passages (extractive QnA).
- **RoBERTa:** A robustly optimized BERT variant, often achieves higher accuracy.
- **DistilBERT:** A smaller, faster version of BERT, suitable for real-time applications.
- **ALBERT, ELECTRA:** Other efficient transformer models for QnA.
- **T5, GPT-3/4:** Can perform both extractive and generative QnA (can generate answers, not just extract).


### Capabilities and Limitations


**Capabilities:**
- Can find precise answers in large documents or datasets.
- Understands context, synonyms, and paraphrased questions.
- Supports multiple languages (with multilingual models).
- Can be fine-tuned for domain-specific QnA (e.g., legal, medical).


**Limitations:**
- Extractive models can only return text spans from the input, not generate new answers.
- May struggle with ambiguous, multi-hop, or reasoning-heavy questions.
- Sensitive to input phrasing and context window size.
- Generative models (like GPT) can hallucinate or make up answers if not carefully prompted.



Let's see a demo of QnA using Hugging Face Transformers!

```python
# Demo: Extractive Question Answering with BERT
from transformers import pipeline
```

```python
# Create a question-answering pipeline
qa_pipeline = pipeline('question-answering', model='distilbert-base-uncased-distilled-squad')
```

```python
context = (
    "Transformers are a type of deep learning model that use self-attention mechanisms to process input data. "
    "They have revolutionized natural language processing tasks such as translation, summarization, and question answering. "
    "BERT, RoBERTa, and GPT are popular transformer models."
    )

question = "What tasks have transformers revolutionized?"

result = qa_pipeline(question=question, context=context)

print('Question:', question)
print('Answer:', result['answer'])
```

```python
# More QnA Examples with the Same Pipeline

examples = [
    {
        'context': (
            "The transformer architecture was introduced in 2017 by Vaswani et al. "
            "It relies on self-attention mechanisms and has become the foundation for models like BERT, GPT, and T5."
        ),
        'question': "Who introduced the transformer architecture?"
    },
    {
        'context': (
            "BERT is a transformer-based model that is pre-trained on large text corpora. "
            "It is widely used for tasks such as question answering, sentiment analysis, and named entity recognition."
        ),
        'question': "What are some tasks BERT is used for?"
    },
    {
        'context': (
            "GPT-3 is a large language model developed by OpenAI. "
            "It can generate text, translate languages, and answer questions in a conversational manner."
        ),
        'question': "Who developed GPT-3?"
    }
]

for ex in examples:
    result = qa_pipeline(question=ex['question'], context=ex['context'])
    print('Question:', ex['question'])
    print('Answer:', result['answer'])
    print('-' * 40)
```

### Limitations of Transformer Models for QnA

Even though transformer models like BERT, RoBERTa, and GPT have greatly improved QnA, they still have important limitations:

1. **Context Window Size:**
   - Models can only "see" a limited number of tokens at once (e.g., 512 for BERT, 2048–4096 for many GPTs). Long documents may be truncated, missing key information.

2. **Extractive vs. Generative:**
   - Most QnA models are extractive—they can only return exact text spans from the input, not generate new answers or explanations.
   - Generative models (like GPT) can make up answers, but may hallucinate or provide incorrect information.

3. **Ambiguity and Reasoning:**
   - Struggle with ambiguous questions, multi-hop reasoning (answering questions that require combining information from multiple places), or questions needing world knowledge.

4. **Domain Adaptation:**
   - Out-of-the-box models may not perform well on specialized domains (e.g., legal, medical) without fine-tuning on domain-specific data.

5. **Sensitivity to Phrasing:**
   - The answer may change if the question is reworded, even if the meaning is the same.

6. **Lack of Up-to-date Knowledge:**
   - Models are only as current as their training data. They cannot answer questions about events or facts after their last update.

7. **Bias and Fairness:**
   - May reflect biases present in training data, leading to unfair or inappropriate answers.

**Summary:**
- Transformers are powerful for QnA, but not perfect. For critical applications, always validate answers and consider human review.

#### Demo: Limitation #2 - Extractive vs. Generative QnA

- Extractive model: can only return exact text spans from the context

```python
context = (
    "The capital of France is Paris. Paris is known for the Eiffel Tower and its rich history. "
    "It is a popular tourist destination."
    )

question1 = "What is the capital of France?"
question2 = "What is the most famous landmark in Paris?"
question3 = "Why do people visit Paris?"

print('Extractive QnA (DistilBERT):')
for q in [question1, question2, question3]:
    result = qa_pipeline(question=q, context=context)
    print('Q:', q)
    print('A:', result['answer'])
    print('-' * 30)
```

```python
# Generative model (if available): can generate new answers, but may hallucinate

try:
    gen_qa = pipeline('text2text-generation', model='google/flan-t5-small')
    print('\nGenerative QnA (Flan-T5):')
    for q in [question1, question2, question3]:
        prompt = f"question: {q}  context: {context}"
        answer = gen_qa(prompt, max_length=50)[0]['generated_text']
        print('Q:', q)
        print('A:', answer)
        print('-' * 30)
except Exception as e:
    print('Generative model not available or failed to load:', e)
```

#### Demo: Limitation #3 - Ambiguity and Multi-hop Reasoning

- Example 1: Ambiguous question

```python
context1 = (
    "Apple was founded by Steve Jobs and Steve Wozniak in California. "
    "Apple is also a popular fruit that is grown in many countries."
    )

question1 = "Who founded Apple?"

result1 = qa_pipeline(question=question1, context=context1)

print('Ambiguous QnA:')
print('Q:', question1)
print('A:', result1['answer'])
print('-' * 30)
```

- Example 2: Multi-hop reasoning required

```python
context2 = (
    "Marie Curie discovered radium. Radium is used in cancer treatment. "
    "Cancer treatment helps save lives."
    )

question2 = "Who discovered something used in cancer treatment?"

result2 = qa_pipeline(question=question2, context=context2)

print('Multi-hop Reasoning QnA:')
print('Q:', question2)
print('A:', result2['answer'])
print('-' * 30)
```

#### Why did the model succeed in these simple ambiguous and multi-hop QnA cases?

- **Ambiguous QnA:**
  - The context included both the company "Apple" and the fruit, but only the company was described as "founded by" people. The model used this clue to pick the right answer (Steve Jobs and Steve Wozniak).
  - If the context or question were less clear, or if both meanings fit, the model might get confused.

- **Multi-hop Reasoning:**
  - The context was short and the reasoning chain was direct: "Marie Curie discovered radium. Radium is used in cancer treatment." The model could connect these facts because they were close together and the logic was simple.
  - In longer or more complex documents, or if the answer required combining information from distant parts, the model would often fail.

**Summary:**
- These demos work because the examples are clear, short, and the answer is easy to extract.
- For more complex, ambiguous, or multi-step reasoning questions, transformer QnA models are much more likely to make mistakes or miss the answer.

```python
# Challenging multi-hop reasoning example
context_challenge = (
    "Alan Turing developed the concept of a Turing machine. "
    "The Turing machine is a fundamental model in computer science. "
    "Computer science is essential for the development of artificial intelligence. "
    "Artificial intelligence is used in self-driving cars."
    )

question_challenge = "Who contributed to something that is used in self-driving cars?"

result_challenge = qa_pipeline(question=question_challenge, context=context_challenge)

print('Challenging Multi-hop QnA:')
print('Q:', question_challenge)
print('A:', result_challenge['answer'])
```

**Expected Reasoning (for humans):**
- Alan Turing → Turing machine → computer science → artificial intelligence → self-driving cars.
- The answer should be "Alan Turing" because he contributed to the Turing machine, which is foundational to the chain leading to self-driving cars.

**What usually happens:**
- Most extractive QnA models will fail to connect all these steps and may return an incomplete or irrelevant answer (e.g., "Turing machine" or "artificial intelligence").
- This demonstrates the limitation: the model cannot reliably perform complex, multi-step reasoning across distant context spans.

#### Limitation #4: Domain Adaptation


**What is it?**

- Out-of-the-box transformer QnA models are trained on general-purpose data (like Wikipedia, news, and web text).

- They may not perform well on specialized domains (e.g., legal, medical, finance, technical manuals) because they lack domain-specific vocabulary, concepts, and reasoning patterns.

**Why does this matter?**

- In real-world business and research, many QnA tasks require understanding of specialized language and context.
- Without fine-tuning on domain-specific data, models may give incorrect, incomplete, or nonsensical answers.

**Let's see a demo:**

- We'll ask a medical question using a general-purpose QnA model and see how it performs.

```python
# Domain adaptation failure demo: Medical QnA with a general model
context_medical = (
    "A myocardial infarction, commonly known as a heart attack, occurs when blood flow to the heart muscle is blocked. "
    "Symptoms include chest pain, shortness of breath, and nausea. Immediate treatment is critical to restore blood flow and minimize heart damage."
    )

question_medical = "What is the first-line treatment for myocardial infarction?"

result_medical = qa_pipeline(question=question_medical, context=context_medical)

print('Medical QnA (General Model):')
print('Q:', question_medical)
print('A:', result_medical['answer'])
```

**Expected (Domain Expert) Answer:**
- The first-line treatment for myocardial infarction typically includes administration of aspirin, nitroglycerin, and rapid revascularization (e.g., percutaneous coronary intervention).
- The context above does not mention these treatments, so a general-purpose model will likely fail or give a vague answer.

**What usually happens:**
- The model may repeat part of the context (e.g., "restore blood flow"), or provide an incomplete or non-specific answer.
- This shows why domain adaptation (fine-tuning on medical data) is crucial for specialized QnA tasks.

**Summary:**
- For accurate answers in specialized fields, transformer models need to be fine-tuned on domain-specific data. General models are not reliable for expert-level QnA.

#### Would OpenAI LLMs (e.g., GPT-4) Perform Differently on Domain QnA?

**Short answer:** Yes, but with important caveats.

**Why OpenAI LLMs might do better:**

- OpenAI LLMs (like GPT-3/4) are trained on much larger and more diverse datasets, including some domain-specific content (e.g., legal, medical, technical documents).

- They are generative, so they can synthesize answers even if the context is missing details.
- They can sometimes "reason" or fill in gaps using their broad world knowledge.
- With careful prompting (e.g., "You are a medical expert..."), they may provide more accurate or complete answers.

**But there are still limitations:**

- Even OpenAI LLMs are not a substitute for true domain expertise or up-to-date, authoritative sources.
- They can still hallucinate, make up facts, or provide plausible-sounding but incorrect answers.
- For critical use cases (e.g., medicine, law), always validate with a human expert or trusted source.

**Summary:**

- OpenAI LLMs often outperform smaller, general-purpose models on domain QnA, but are not perfect.
- Fine-tuned, domain-specific models or retrieval-augmented systems are best for expert-level accuracy.

#### Limitation #5: Sensitivity to Phrasing


**What is it?**

- Transformer QnA models can give different answers to questions with the same meaning but different wording.
- Small changes in phrasing, synonyms, or question structure can lead to different or even incorrect answers.

**Why does this matter?**

- In real-world applications, users may ask the same question in many ways.
- A robust QnA system should be consistent and reliable, regardless of phrasing.

**Let's see a demo:**

- We'll ask several versions of the same question and compare the answers.

```python
# Sensitivity to phrasing demo
context_phrasing = (
    "The tallest mountain in the world is Mount Everest, which is located in the Himalayas on the border between Nepal and China. "
    "It has an official elevation of 8,848.86 meters (29,031.7 ft)."
    )
```

```python
questions_phrasing = [
    "What is the tallest mountain in the world?",
    "Which mountain has the highest elevation?",
    "Name the highest peak on Earth.",
    "Which is the world's tallest mountain?",
    "What mountain is 8,848.86 meters tall?"
    ]

print('Sensitivity to Phrasing (DistilBERT):')
for q in questions_phrasing:
    result = qa_pipeline(question=q, context=context_phrasing)
    print('Q:', q)
    print('A:', result['answer'])
    print('-' * 30)
```

**What to look for:**

- Ideally, the model should answer "Mount Everest" for all versions of the question.
- In practice, you may see inconsistent answers, partial answers, or even failures to answer correctly, depending on the phrasing.

**Summary:**

- Transformer QnA models are sensitive to how questions are asked.
- For robust applications, consider using question normalization, paraphrasing, or retrieval-augmented techniques to improve consistency.

**Observation:**

- In this simple factual example, the model gave a consistent and correct answer ("Mount Everest") for all phrasings.
- This shows that for well-known facts and clear contexts, transformer QnA models can be robust to rephrasing.


**However:**

- For more complex, ambiguous, or less common questions, or in specialized domains, you may still see inconsistent or incorrect answers.
- Always test with a variety of question forms, especially for critical or business applications.

#### Limitation #6: Lack of Up-to-date Knowledge

**What is it?**
- Transformer models (like BERT, GPT, T5) are trained on large datasets, but their knowledge is fixed at the time of training.
- They cannot access new information, events, or facts that occurred after their last training update.

**Why does this matter?**
- In fast-changing fields (news, finance, technology, regulations), models may give outdated or incorrect answers.
- For business, legal, or scientific use, relying on up-to-date information is critical.

**Example:**
- If a model was trained in 2021, it cannot answer questions about events in 2022 or later.
- It may not know about new products, laws, scientific discoveries, or recent world events.

#### Limitation #7: Bias and Fairness

**What is it?**
- Transformer models learn from large datasets that reflect real-world language, including stereotypes, biases, and imbalances.
- As a result, models can produce biased, unfair, or inappropriate answers, even if not intended by the user.

**Why does this matter?**
- In business, education, healthcare, and public-facing applications, biased outputs can cause harm, reinforce stereotypes, or lead to unfair decisions.
- Regulatory and ethical standards increasingly require AI systems to be fair, transparent, and accountable.

**Examples of Bias:**
- Gender bias: Assuming certain professions or roles are associated with a specific gender.
- Racial or cultural bias: Producing answers that reflect stereotypes or underrepresent certain groups.
- Confirmation bias: Repeating or amplifying common but incorrect beliefs from the training data.

**Demo:**
- Let's see how a QnA model might reflect bias in its answers.

```python
# Demo: Bias and Fairness in QnA
contexts_bias = [
    # Gender bias example
    ("The doctor examined the patient and prescribed medication. The nurse assisted with the procedure.", "Who is the nurse?"),
    # Racial/cultural bias example
    ("John is an engineer from the United States. Priya is an engineer from India. Both are experts in their field.", "Who is better at engineering?"),
    # Confirmation bias example
    ("Vaccines are a safe and effective way to prevent disease. Some people believe vaccines cause autism, but scientific studies have found no link.", "Do vaccines cause autism?")
]

print('Bias and Fairness Demo:')
for context, question in contexts_bias:
    result = qa_pipeline(question=question, context=context)
    print('Q:', question)
    print('A:', result['answer'])
    print('-' * 30)
```

**Analysis: Bias and Fairness Demo Results**

Let's review the model's answers and what they reveal about bias and fairness in transformer QnA systems:

| Question | Model's Answer | Bias/Fairness Issue | Notes |
|---|---|---|---|
| Who is the nurse? | The doctor | ⚠️ Gender/role bias | The model incorrectly assumes the nurse and doctor are the same, possibly due to lack of gender cues or context. In real-world data, models may also default to gender stereotypes (e.g., nurse=female, doctor=male). |
| Who is better at engineering? | Priya | ⚠️ Cultural/national bias | The model picks one name (Priya) without evidence, possibly reflecting bias or random selection. There is no basis in the context for this answer. |
| Do vaccines cause autism? | Some people believe vaccines cause autism | ⚠️ Confirmation bias | The model repeats a common misconception, even though the context says studies found no link. It does not clearly state the scientific consensus. |

**Key Observations:**
- The model can reflect or amplify biases present in its training data or context, even when the context is neutral or clarifies the correct answer.
- For ambiguous or stereotype-prone questions, the model may default to common associations or make unsupported assumptions.
- On controversial topics, the model may hedge or repeat misinformation, rather than assert the correct, evidence-based answer.

**Takeaway:**
- Always review QnA model outputs for bias, fairness, and accuracy—especially in sensitive or high-stakes domains.
- Consider using additional safeguards: prompt engineering, post-processing, human review, or bias mitigation techniques.
- Responsible AI requires ongoing monitoring and improvement to ensure fairness and trustworthiness.

#### Can Transformers Do Text-to-Image Generation?
---

Yes, but with important differences from text-based tasks! 

#### 1. What is Text-to-Image Generation?
- **Text-to-image** means creating a new image from a written description (prompt).
- Example: Given the prompt _"A cat wearing sunglasses on a beach"_, the model generates a unique image matching that description.

#### 2. How Do Transformers Fit In?
- **Transformers** revolutionized language tasks (like translation, QnA, summarization).
- For images, researchers adapted transformer architectures to handle visual data, either alone or combined with other techniques (like diffusion models).
- **Key models:**
  - **DALL·E** (OpenAI): Uses transformers to generate images from text prompts.
  - **Stable Diffusion** (Stability AI): Combines transformers with diffusion processes for high-quality, fast image generation.
  - **Imagen** (Google): Uses large transformer models for text-to-image.

#### 3. How Is This Different from Text Generation?
- **Text generation**: Output is a sequence of words/tokens.
- **Text-to-image**: Output is a grid of pixels (an image), which is much more complex!
- Models must "understand" both language and how to represent visual concepts.

#### 4. How Does It Work (Simplified)?
1. **Text Encoding**: The prompt is converted into a numerical representation using a text transformer (like T5 or CLIP).
2. **Image Generation**: Another model (often a diffusion model or a transformer-based decoder) takes this representation and generates an image step-by-step.
3. **Training**: These models are trained on huge datasets of images paired with captions, learning to "imagine" what a prompt should look like.

#### 5. What Tools Can I Use?
- **Hugging Face Diffusers**: Python library for text-to-image models (DALL·E, Stable Diffusion, etc.).
- **APIs**: Many companies offer cloud APIs for text-to-image (OpenAI, Stability AI, etc.).

#### 6. Example Prompts
- "A futuristic cityscape at sunset, in watercolor style."
- "A dog riding a skateboard, photorealistic."

#### 7. Limitations & Considerations
- **Computationally intensive**: Generating images requires more resources than text.
- **Biases**: Models can reflect biases in their training data.
- **Not perfect**: Sometimes outputs are strange or don't match the prompt exactly.

**In summary:**
- Transformers can be used for text-to-image, but require special architectures and lots of data.
- This is a cutting-edge area of GenAI, opening up creative and business possibilities!

#### Demo: Image Captioning with Transformers
---

**Image captioning** is the task of generating a natural language description for a given image. Transformer-based models can "see" an image and describe it in words, combining computer vision and natural language processing.

**How Does It Work?**
- The model takes an image as input, processes it with a vision encoder (like a Vision Transformer or CNN), and then uses a language decoder (transformer) to generate a caption.
- Popular models: BLIP, OFA, ViT-GPT2, and others (available via Hugging Face `transformers` and `diffusers`).

**What You'll See in the Demo**
- Load a sample image.
- Use a pre-trained transformer model to generate a caption.
- Display the image and the generated caption.



```python
# Install required libraries (if not already installed)
# !pip install transformers pillow requests

from transformers import BlipProcessor, BlipForConditionalGeneration
from PIL import Image
import requests
import matplotlib.pyplot as plt
```

```python
# Download a sample image (you can replace this URL with any image you like)
img_url = "https://images.unsplash.com/photo-1519125323398-675f0ddb6308"
image = Image.open(requests.get(img_url, stream=True).raw)
```

```python
# Load BLIP processor and model (pretrained for image captioning)
processor = BlipProcessor.from_pretrained("Salesforce/blip-image-captioning-base")
model = BlipForConditionalGeneration.from_pretrained("Salesforce/blip-image-captioning-base")
```

```python
# Preprocess the image and generate caption
inputs = processor(image, return_tensors="pt")
out = model.generate(**inputs)
caption = processor.decode(out[0], skip_special_tokens=True)
```

```python
# Show the image and the generated caption
plt.imshow(image)
plt.axis('off')
plt.title(f"Caption: {caption}")
plt.show()

print("Generated Caption:", caption)
```

# Zero-Shot Prompting with Large Language Models

Zero-shot prompting refers to the technique of using a pre-trained language model, 
such as GPT-4, to perform tasks without providing any specific examples or fine-tuning. 

Instead, you simply give the model an instruction or question, 
and it generates a response based on its training and knowledge.

## Key Features of Zero-Shot Prompting

- **Direct Instructions**: You provide a clear instruction or question.
- **Generalization**: The model uses its vast training data to infer a suitable response.
- **Versatility**: Applicable to various tasks, including text generation, question answering, summarization, and more.
- **No Fine-Tuning Required**: Works out-of-the-box with the model's existing knowledge.

#### let us define a function to get answer for a qs

````python
def question_answering(question, context):

    prompt   = f"Q: {question}\nContext: {context}\nA:"
    
    # Generate an answer using GPT-3.5
    response = client.chat.completions.create(
            model      = "gpt-4o-mini",
            messages   = [
                {"role": "system", "content": 'You are a helpful assistant in the field of medical domain'},
                {"role": "user",   "content": f'{prompt}'}
            ],
            max_tokens = 50, 
            n          = 1,
            temperature= 0.001
        )

    # Extract the answer from the API response
    answer = response.choices[0].message.content.strip()

    return answer
````

### text classification

````python
def get_completion(prompt, model="gpt-3.5-turbo"):
    
    messages = [{"role": "user", "content": prompt}]
    
    response = client.chat.completions.create(
        model      = model,
        messages   = messages,
        temperature= 0, # this is the degree of randomness of the model's output, very deterministic
    )
    return response.choices[0].message.content
````

**Ex 01 (text)**

```python
response = client.chat.completions.create(
                messages=[
                    {"role": "user", "content": "what is Data Science"}
                ],
                model="gpt-4o-mini",
                max_tokens = 100
)
```

```python
type(response)
```

Response Object Structure

| Field                                 | Description                                         | Example                               |
| ------------------------------------- | --------------------------------------------------- | ------------------------------------- |
| `response.id`                         | Unique ID for the completion call                   | `'chatcmpl-abc123'`                   |
| `response.object`                     | Always `'chat.completion'`                          | `'chat.completion'`                   |
| `response.created`                    | Timestamp (UNIX) when the completion was created    | `1720000000`                          |
| `response.model`                      | Model used                                          | `'gpt-4o-mini'`                       |
| `response.choices`                    | List of possible completions (usually just 1)       | List of `ChatCompletionChoice`        |
| `response.choices[0].message.role`    | Role of the message in the response (`'assistant'`) | `'assistant'`                         |
| `response.choices[0].message.content` | Main generated text/content                         | `'Data Science is the field that...'` |
| `response.choices[0].finish_reason`   | Why the model stopped (e.g., `'stop'`, `'length'`)  | `'stop'`                              |
| `response.usage.prompt_tokens`        | Tokens used in prompt                               | `7`                                   |
| `response.usage.completion_tokens`    | Tokens used in output                               | `25`                                  |
| `response.usage.total_tokens`         | Total tokens used                                   | `32`                                  |


```python
# Text response
answer = response.choices[0].message.content
print(answer)
```

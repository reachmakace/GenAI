These examples explains how to use AutoTokenizer transformer library to load an pretrained model to create get the embeddings. Various pretrained models are used in this file for demostratin

```python
# AutoTokenizer is a class from the Hugging Face Transformers library.
# AutoTokenizer is a smart tokenizer loader. You give it the name of a pretrained model, and it automatically loads the correct tokenizer for that model.

from transformers import AutoTokenizer
```

```python
# Dummy sentence
text = "Transformers library provides a variety of tokenization techniques!"
```

1. WordPiece Tokenizer (used by BERT)

```python
# Upon loading a tokenizer using AutoTokenizer, it will store the token vocabulary and the tokenization method used by the specified model. 
# It will be store in the local cache directory for future use. C:\Users\reach\.cache\huggingface\hub
print("=== WordPiece Tokenizer ===")
bert_tokenizer = AutoTokenizer.from_pretrained("bert-base-uncased")
bert_tokens    = bert_tokenizer.tokenize(text)
bert_encoded   = bert_tokenizer.encode(text)
print("Tokens:", bert_tokens)
print("Encoded:", bert_encoded)
print("Decoded:", bert_tokenizer.decode(bert_encoded))
```

2. Byte Pair Encoding (BPE) Tokenizer (used by GPT-2)

```python
print("\n=== Byte Pair Encoding (BPE) Tokenizer ===")
gpt2_tokenizer = AutoTokenizer.from_pretrained("gpt2")
gpt2_tokens = gpt2_tokenizer.tokenize(text)
gpt2_encoded = gpt2_tokenizer.encode(text)
print("Tokens:", gpt2_tokens)
print("Encoded:", gpt2_encoded)
print("Decoded:", gpt2_tokenizer.decode(gpt2_encoded))
```

3. Unigram Tokenizer (used by Albert)

```python
print("\n=== Unigram Tokenizer ===")
albert_tokenizer = AutoTokenizer.from_pretrained("albert-base-v2")
albert_tokens = albert_tokenizer.tokenize(text)
albert_encoded = albert_tokenizer.encode(text)
print("Tokens:", albert_tokens)
print("Encoded:", albert_encoded)
print("Decoded:", albert_tokenizer.decode(albert_encoded))
```

4. SentencePiece Tokenizer (used by T5)

```python

print("\n=== SentencePiece Tokenizer ===")
t5_tokenizer = AutoTokenizer.from_pretrained("t5-small")
t5_tokens = t5_tokenizer.tokenize(text)
t5_encoded = t5_tokenizer.encode(text)
print("Tokens:", t5_tokens)
print("Encoded:", t5_encoded)
print("Decoded:", t5_tokenizer.decode(t5_encoded))
```

### Natural Language Inference (NLI) 
- is a common task in natural language understanding where you determine the relationship between two given sentences: "premise" and "hypothesis."
-------------------

```python
from sentence_transformers import SentenceTransformer, util
```

```python
import os
os.environ["SENTENCE_TRANSFORMERS_HOME"] = r'D:\AI-DATASETS\07-Hugging-Face-Data\sentence-transformers'
```

```python
# Load a pre-trained Sentence Transformer model
#model_name = "paraphrase-MiniLM-L6-v2"
#model_name = "bert-base-uncased"
model_name = "bert-base-nli-mean-tokens"

embedder   = SentenceTransformer(model_name)
```

```python
# Example premise and hypothesis sentences
premise    = "A cat is playing outside."
hypothesis = "A feline is outdoors."
```

```python
# Embed the premise and hypothesis
embed_premise    = embedder.encode(premise, convert_to_tensor=True)
embed_hypothesis = embedder.encode(hypothesis, convert_to_tensor=True)
```

```python
# Calculate cosine similarity between the embeddings
similarity_score = util.pytorch_cos_sim(embed_premise, embed_hypothesis)
```

```python
# Set a threshold for NLI task (you can define your own threshold)
threshold = 0.7
```

```python
# Check if the similarity score exceeds the threshold
if similarity_score > threshold:
    print("Entailment: The hypothesis can be inferred from the premise.")
else:
    print("Contradiction: The hypothesis contradicts the premise.")
```

```python
from sentence_transformers import SentenceTransformer, util
from transformers import pipeline
import spacy
```

```python
import os
#os.environ["SENTENCE_TRANSFORMERS_HOME"] = r'D:\AI-DATASETS\07-Hugging-Face-Data\sentence-transformers'
```

```python
# Load a pre-trained Sentence Transformer model
model_name = "paraphrase-MiniLM-L6-v2"
embedder = SentenceTransformer(model_name)
```

```python
# Load a pre-trained NER model (Hugging Face Transformers)
nlp_ner = pipeline("ner", model="dbmdz/bert-large-cased-finetuned-conll03-english")
```

```python
# Sample text
text = "Apple Inc. is a technology company based in Cupertino, California. $10m dollars"
```

```python
# Encode the text using Sentence Transformer
embeddings = embedder.encode(text, convert_to_tensor=True)
```

```python
# Convert sentences to text for NER
sentences_text = [text]
```

```python
# Perform NER on the encoded text
ner_results = nlp_ner(sentences_text)
```

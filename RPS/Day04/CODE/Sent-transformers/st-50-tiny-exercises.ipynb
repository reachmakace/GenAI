#### Semantic Similarity Finder

=== Instructions ===

In this exercise, you will:

1. Encode a list of sentences into embeddings.
2. Compute cosine similarity between all pairs.
3. Find which two sentences are most similar.

> install necessary libraries if needed

```python
from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity
import numpy as np
```

```python
sentences = [
    # Technology
    "Apple releases the new iPhone with improved battery life.",
    "Samsung unveils its latest Galaxy smartphone.",
    "Google announces breakthrough in quantum computing.",
    "Microsoft launches a new version of Windows.",
    "The latest laptop model offers faster performance.",
    "New smartwatch features advanced health tracking.",

    # Fitness & Health
    "Daily exercise improves overall health and fitness.",
    "Morning yoga helps with stress relief and flexibility.",
    "Running regularly boosts heart health.",
    "A balanced diet is essential for good health.",
    "Strength training builds muscle and endurance.",
    "Walking 10,000 steps daily improves wellbeing.",

    # Travel & Tourism
    "Paris is famous for the Eiffel Tower and its cafés.",
    "Tokyo offers a mix of tradition and technology.",
    "New York City is known as the city that never sleeps.",
    "Rome attracts tourists with its ancient ruins.",
    "Bali is a popular destination for beaches and temples.",
    "London is famous for its history and landmarks."
]
```

```python
# Load Model (using Sentence transformer, all mini)
model = SentenceTransformer('all-MiniLM-L6-v2')
```

```python
# Encode Sentences (all the sentences above)
embeddings = model.encode(sentences)
```

```python
# Compute Cosine Similarities
similarity_matrix = cosine_similarity(embeddings)
```

```python
# Viz
import seaborn as sns
sns.heatmap(similarity_matrix)
```

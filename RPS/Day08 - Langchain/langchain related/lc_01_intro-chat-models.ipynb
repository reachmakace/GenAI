#### Chatmodels integration 
------------------------------------

- **Large Language Models (LLMs)**:
  - Powerful machine learning models adept at various language-related tasks, including:
    - Text generation
    - Translation
    - Summarization
    - Question answering
  - Capable of performing these tasks without specific fine-tuning for each scenario.

- **Chat Model Interface**:
  - Commonly used to interact with modern LLMs.
  - Accepts a list of messages as input and returns a single message as output.

- **New Features in Chat Models**:
  - **Tool Calling**:
    - Many chat models now support a native tool-calling API.
    - Allows the model to interact with external services, APIs, and databases.
    - Useful for extracting structured information from unstructured data.
    - Enables the creation of advanced applications with AI-driven interactions.

  - **Structured Output**:
    - Enables chat models to respond in specific formats, such as JSON.
    - Useful for producing outputs that follow a predefined schema.

  - **Multimodality**:
    - Extends the model’s capabilities beyond text processing.
    - Supports handling of diverse data types, including images, audio, and video, broadening the range of tasks it can perform.

```python
# user question
# This is the input you want to dynamically insert into your prompt.
question = "Which NFL team won the Super Bowl in the 2010 season?"
```

```python
# This is a prompt template used for passing instructions to an LLM (like OpenAI GPT).
# {question} is a placeholder (variable) that will be replaced later by actual text.

template = """

Question: {question}

Answer: 

"""
```

```python
# Creating the PromptTemplate object
prompt = PromptTemplate(
        template       = template,
        input_variables= ['question']  # tells the template which fields can be dynamically replaced.
)
```

### Models in LangChain: LLMs, Chat Models, and Embedding Models

LangChain provides interfaces for different types of models:
- **LLMs (Large Language Models):** For text generation and completion tasks.
- **Chat Models:** For conversational AI, handling multi-turn dialogues.
- **Embedding Models:** For converting text into vector representations for similarity search and retrieval.

Below are example usages for each model type.

```python
# Example: Using a Chat Model for conversation
from langchain_openai import ChatOpenAI

chat_model = ChatOpenAI(model_name="gpt-3.5-turbo")
messages = [
    {"role": "system", "content": "You are a helpful assistant."},
    {"role": "user", "content": "Tell me a joke."}
]
response = chat_model.invoke(messages)
response
```

```python
# Example: Using an Embedding Model for text embeddings
from langchain_openai import OpenAIEmbeddings

embedding_model = OpenAIEmbeddings()

text = "LangChain enables LLM-powered applications."
embedding = embedding_model.embed_query(text)
print(embedding)
```

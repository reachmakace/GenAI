## Basic Text Generation

Start with fundamental text generation techniques.

**Key Concept**: Prompt engineering is like writing clear requirements - the better your prompt, the better the output.

```python
# SIMPLE TEXT GENERATOR

def create_basic_text_generator(content_type="article"):
    """Create a basic text generator for different content types"""
    
    templates = {
        "article": """
        Write a well-structured article about: {topic}
        
        Requirements:
        - Include an engaging introduction
        - Organize content with clear sections
        - Provide practical insights
        - End with a compelling conclusion
        
        Article:
        """,
        
        "blog_post": """
        Create an engaging blog post about: {topic}
        
        Style:
        - Conversational and friendly tone
        - Include personal insights or examples
        - Make it shareable and engaging
        - Add a call-to-action at the end
        
        Blog Post:
        """,
        
        "technical_doc": """
        Write technical documentation for: {topic}
        
        Requirements:
        - Clear and precise language
        - Include code examples where relevant
        - Organize with headers and bullet points
        - Focus on practical implementation
        
        Technical Documentation:
        """
    }
    
    template = templates.get(content_type, templates["article"])
    prompt   = PromptTemplate(template=template, input_variables=["topic"])
    
    return LLMChain(llm=llm, prompt=prompt)
```

```python
# Test different content types
test_topics = {
    "article":       "The Future of Artificial Intelligence in Software Development",
    "blog_post":     "5 Java Performance Tips Every Developer Should Know",
    "technical_doc": "RESTful API Design Best Practices"
}

for content_type, topic in test_topics.items():
    print(f"\n Generating {content_type.upper()} on: {topic}")
    generator = create_basic_text_generator(content_type)
    result = generator.run(topic=topic)
    
    # Show first 200 characters as preview
    preview = result[:200] + "..." if len(result) > 200 else result
    print(f" Preview: {preview}")
    print(f" Generated {len(result)} characters ({len(result.split())} words)")
    print("-" * 40)
```

```python
# TEMPLATE-BASED GENERATION WITH PARAMETERS

def create_parametric_generator(template_name):
    """Create generators with multiple input parameters for more control"""
    
    templates = {
        "project_proposal": """
        Create a project proposal with the following details:
        
        Project: {project_name}
        Budget: {budget}
        Timeline: {timeline}
        Team Size: {team_size}
        Key Objectives: {objectives}
        
        Generate a comprehensive project proposal including:
        - Executive Summary
        - Project Scope and Objectives
        - Timeline and Milestones
        - Budget Breakdown
        - Team Structure
        - Risk Assessment
        
        Project Proposal:
        """,
        
        "product_description": """
        Create a compelling product description:
        
        Product: {product_name}
        Category: {category}
        Target Audience: {audience}
        Key Features: {features}
        Price Range: {price_range}
        
        Generate an engaging product description that includes:
        - Compelling headline
        - Problem it solves
        - Key benefits and features
        - Social proof or credibility
        - Call to action
        
        Product Description:
        """,
        
        "meeting_agenda": """
        Create a structured meeting agenda:
        
        Meeting: {meeting_title}
        Duration: {duration}
        Attendees: {attendees}
        Objectives: {objectives}
        
        Generate a professional meeting agenda with:
        - Meeting overview
        - Time-boxed agenda items
        - Discussion points
        - Action items template
        - Next steps
        
        Meeting Agenda:
        """
    }
    
    template = templates[template_name]
    # Extract variable names from template
    import re
    variables = re.findall(r'\{(\w+)\}', template)
    
    prompt = PromptTemplate(template=template, input_variables=variables)
    return LLMChain(llm=llm, prompt=prompt), variables

# Example: Generate a project proposal
proposal_generator, variables = create_parametric_generator("project_proposal")
print(f" Project Proposal Generator requires: {variables}")

# Generate with specific parameters
proposal = proposal_generator.run(
    project_name="AI-Powered Customer Support System",
    budget="$150,000",
    timeline="6 months",
    team_size="5 developers + 1 PM",
    objectives="Reduce support ticket resolution time by 40% using NLP and automation"
)

print(f"\n📄 Generated Project Proposal ({len(proposal)} characters):")
print(proposal[:300] + "...\n")

# Example: Product description
product_generator, variables = create_parametric_generator("product_description")
print(f" Product Description Generator requires: {variables}")

product_desc = product_generator.run(
    product_name="CodeMaster Pro",
    category="Developer Tools",
    audience="Java developers and software architects",
    features="AI code completion, automated refactoring, performance analysis",
    price_range="$29-99/month"
)

print(f"\n Generated Product Description:")
print(product_desc[:250] + "...")
print("-" * 50)
```

## Code Generation for Developers

Specialized generation techniques for software development tasks.


```python
# 1. JAVA CODE GENERATION

def create_java_code_generator(code_type="class"):
    """Generate Java code with proper patterns and conventions"""
    
    templates = {
        "class": """
        Generate a Java class with the following specifications:
        
        Class Name: {class_name}
        Purpose: {purpose}
        Fields: {fields}
        Methods: {methods}
        
        Requirements:
        - Follow Java naming conventions
        - Include proper encapsulation (private fields, public getters/setters)
        - Add JavaDoc comments
        - Include proper imports
        - Follow best practices
        
        Java Class:
        """,
        
        "service": """
        Generate a Java service class following Spring Boot patterns:
        
        Service: {service_name}
        Domain: {domain}
        Operations: {operations}
        Dependencies: {dependencies}
        
        Requirements:
        - Use @Service annotation
        - Implement dependency injection
        - Include error handling
        - Add logging
        - Follow SOLID principles
        
        Spring Service Class:
        """,
        
        "rest_controller": """
        Generate a Java REST controller:
        
        Controller: {controller_name}
        Base Path: {base_path}
        Endpoints: {endpoints}
        Request/Response Types: {data_types}
        
        Requirements:
        - Use Spring Boot annotations (@RestController, @RequestMapping, etc.)
        - Include proper HTTP methods and status codes
        - Add validation and error handling
        - Include API documentation comments
        - Follow REST conventions
        
        REST Controller:
        """
    }
    
    template = templates.get(code_type, templates["class"])
    variables = re.findall(r'\{(\w+)\}', template)
    prompt = PromptTemplate(template=template, input_variables=variables)
    
    return LLMChain(llm=llm, prompt=prompt)

# Generate a Java class
class_generator = create_java_code_generator("class")
java_class = class_generator.run(
    class_name="UserAccount",
    purpose="Represents a user account in the system",
    fields="id (Long), username (String), email (String), createdDate (LocalDateTime), isActive (boolean)",
    methods="constructors, getters/setters, equals/hashCode, toString, activate(), deactivate()"
)

print(" Generated Java Class:")
print(java_class[:400] + "...\n")

# Generate a Spring service
service_generator = create_java_code_generator("service")
java_service = service_generator.run(
    service_name="UserService",
    domain="User management",
    operations="create user, find user, update user, delete user, authenticate",
    dependencies="UserRepository, PasswordEncoder, EmailService"
)

print("🔧 Generated Spring Service:")
print(java_service[:400] + "...\n")

# Generate REST controller
controller_generator = create_java_code_generator("rest_controller")
java_controller = controller_generator.run(
    controller_name="UserController",
    base_path="/api/users",
    endpoints="GET /users, POST /users, PUT /users/{id}, DELETE /users/{id}, GET /users/{id}",
    data_types="UserDTO, CreateUserRequest, UpdateUserRequest"
)

print(" Generated REST Controller:")
print(java_controller[:400] + "...")
print("-" * 50)
```

```python
# 2. API DOCUMENTATION GENERATOR

def create_api_doc_generator():
    """Generate comprehensive API documentation"""
    
    template = """
    Generate comprehensive API documentation for:
    
    API Name: {api_name}
    Base URL: {base_url}
    Endpoints: {endpoints}
    Authentication: {auth_method}
    
    Create documentation that includes:
    
    ## Overview
    - API description and purpose
    - Base URL and versioning
    - Authentication method
    
    ## Endpoints
    For each endpoint:
    - HTTP method and path
    - Description and purpose
    - Request parameters
    - Request body example
    - Response format and examples
    - Error codes and messages
    
    ## Code Examples
    - cURL examples
    - Java/JavaScript client examples
    
    API Documentation:
    """
    
    prompt = PromptTemplate(
        template=template,
        input_variables=["api_name", "base_url", "endpoints", "auth_method"]
    )
    
    return LLMChain(llm=llm, prompt=prompt)

# Generate API documentation
doc_generator = create_api_doc_generator()
api_docs = doc_generator.run(
    api_name="User Management API",
    base_url="https://api.example.com/v1",
    endpoints="GET /users, POST /users, PUT /users/{id}, DELETE /users/{id}",
    auth_method="Bearer token (JWT)"
)

print("Generated API Documentation:")
# Show structured preview
doc_lines = api_docs.split('\n')[:15]
for line in doc_lines:
    print(line)
print("...")
print(f"\n Total documentation: {len(api_docs)} characters")
print("-" * 50)
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

```python
# Seed for reproducibility
np.random.seed(42)
```

```python
# Generate dummy data
x = np.linspace(0, 10, 30)
y_true = 2 * x + 1
```

```python
# Introduce noise and outliers
noise = np.random.normal(0, 1, 30)
outliers = np.array([0] * 20 + [15, -15, 25, -10, -20, 10, -10, 0, 10, 15])
y_noisy = y_true + noise + outliers
```

```python
# Simple linear regression for prediction
coefficients = np.polyfit(x, y_noisy, 1)
y_pred = np.poly1d(coefficients)(x)
```

```python
# Plotting
plt.figure(figsize=(10, 6))
plt.scatter(x, y_noisy, color='blue', label='True Values with Outliers')
plt.plot(x, y_pred, color='red', label='Predictions')
plt.legend()
plt.title('True vs Predicted Values')
plt.xlabel('X')
plt.ylabel('Y')
plt.grid(True)
```

```python
# Mean Absolute Error (MAE)
mae = mean_absolute_error(y_true, y_pred)
print(f"MAE: {mae}")
```

```python
# Mean Squared Error (MSE)
mse = mean_squared_error(y_true, y_pred)
print(f"MSE: {mse}")
```

```python
# Root Mean Squared Error (RMSE)
rmse = np.sqrt(mse)
print(f"RMSE: {rmse}")
```

```python
# Mean Absolute Percentage Error (MAPE)
mape = np.mean(np.abs((y_true - y_pred) / y_true)) * 100
print(f"MAPE: {mape}%")

# Mean Percentage Error (MPE)
mpe = np.mean((y_true - y_pred) / y_true) * 100
print(f"MPE: {mpe}%")

# R^2 score
r2 = r2_score(y_true, y_pred)
print(f"R2: {r2}")

# Adjusted R^2 score
n = len(y_true)  # number of data points
p = 1  # number of predictors
adj_r2 = 1 - (1 - r2) * (n - 1) / (n - p - 1)
print(f"Adjusted R2: {adj_r2}")
```

### Word embedding 
- using Gensim 

- One of its standout features is the implementation of `Word2Vec` for word embeddings. 
- Word embeddings represent words as dense vectors where semantically similar words are close in the vector space.

```python
# Sample data
sentences = [
    ["I", "love", "machine", "learning"],
    ["machine", "learning", "is", "fun"],
    ["deep", "learning", "is", "a", "subset", "of", "machine", "learning"],
    ["machine", "washing", "powder"],
    ["machine", "washing", "powder"],
    ["machine", "washing", "powder"]
]
```

```python
#pip install gensim
```

```python
from gensim.models import Word2Vec
```

```python
model = Word2Vec(sentences, 
                 vector_size= 10, 
                 window     = 3, 
                 min_count  = 1, 
                 workers    = 4)

model.save("word2vec.model")
```

#### Parameters:

- `vector_size`: Number of dimensions of the embedding.
- `window`: Maximum distance between the current and predicted word within a sentence.
- `min_count`: Ignores all words with total frequency lower than this.
- `workers`: Use these many worker threads to train the model.

```python
vector = model.wv['washing']  # get vector for word 'machine'
vector
```

```python
similar = model.wv.most_similar('washing', topn=5)
similar
```

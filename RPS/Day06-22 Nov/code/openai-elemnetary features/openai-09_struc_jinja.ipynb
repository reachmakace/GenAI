#### basic example

```python
# Basic Jinja2 Template as a String
template_str = "Hello, {{ name }}! Welcome to {{ platform }}."
```

```python
# Data to Inject into the Template
data = {
    "name":     "Alice",
    "platform": "OpenAI"
}
```

```python
# Render the Template with Data
template = Template(template_str)

output   = template.render(data)
```

#### Adding Conditionals (looping)

```python
template_str = """
{% if is_member %}
Hello, {{ name }}! Thanks for being a member.
{% else %}
Hello, {{ name }}! Join us to enjoy more features.
{% endif %}

Your Interests:
{% for interest in interests %}
- {{ interest }}
{% endfor %}
"""
```

```python
data = {
    "name":      "Bob",
    "is_member": True,
    "interests": ["AI", "Data Science", "Machine Learning"]
}
```

```python
template = Template(template_str)
output   = template.render(data)

print(output)
```

#### nested conditionals

```python
template_str = """
Hello, {{ user.name }}!

{% if user.is_member %}
    {% if user.membership_type == 'Premium' %}
        Welcome back, Premium Member! Enjoy your exclusive benefits.
    {% else %}
        Thanks for being a valued member. Consider upgrading to Premium for more features.
    {% endif %}
{% else %}
    You’re not a member yet. Join us to unlock amazing content!
{% endif %}

Your Interests:
{% for interest in user.interests %}
- {{ interest }}
{% endfor %}
"""
```

```python
# Complex Data Structure (Nested Dictionary)
data = {
    "user": {
        "name": "Charlie",
        "is_member": True,
        "membership_type": "Premium",
        "interests": ["Artificial Intelligence", "Deep Learning", "Robotics"]
    }
}
```

```python
# Render the Template
template = Template(template_str)
output = template.render(data)

# Display the Output
print(output)
```

#### Handling Complex Nested Data with Loops & Conditionals

```python
template_str = """
Company: {{ company.name }}

Employees:
{% for employee in company.employees %}
- {{ employee.name }} ({{ employee.position }})
  {% if employee.projects %}
    Projects:
    {% for project in employee.projects %}
      * {{ project.name }} {% if project.is_active %}(Active){% else %}(Completed){% endif %}
    {% endfor %}
  {% else %}
    No projects assigned yet.
  {% endif %}
{% endfor %}
"""
```

```python
data = {
    "company": {
        "name": "Tech Innovations Ltd.",
        "employees": [
            {"name": "Alice", "position": "Data Scientist", "projects": [
                {"name": "AI Chatbot", "is_active": True},
                {"name": "Recommendation System", "is_active": False}
            ]},
            {"name": "Bob", "position": "ML Engineer", "projects": []},
            {"name": "Charlie", "position": "Product Manager", "projects": [
                {"name": "Product Launch", "is_active": True}
            ]}
        ]
    }
}
```

```python
# Render the Template
template = Template(template_str)
output = template.render(data)

print(output)
```

#### looping & OpenAI

```python
# Jinja Template for Dynamic Prompt
template_str = """
Explain the topic: {{ topic }}.

{% if is_advanced %}
Provide an advanced explanation with technical details. 
Cover:
{% for concept in concepts %}
- {{ concept.name }} {% if concept.is_critical %}(Critical){% endif %}
  {% if concept.details %}
    Including: {{ concept.details | join(", ") }}
  {% endif %}
{% endfor %}
{% else %}
Provide a beginner-friendly explanation. 
Focus on:
{% for concept in concepts[:2] %}
- {{ concept.name }}
{% endfor %}
{% endif %}
"""
```

```python
# Create the Jinja Template
template = Template(template_str)
```

```python
# Dynamic Context for Jinja
context = {
    "is_advanced": True,  # Change to False for beginner-level prompts
    "topic":       "Machine Learning",
    "concepts": [
        {"name": "Supervised Learning",   "is_critical": True, "details": ["Regression", "Classification"]},
        {"name": "Unsupervised Learning", "is_critical": False, "details": ["Clustering", "Dimensionality Reduction"]},
        {"name": "Reinforcement Learning", "is_critical": True, "details": ["Agent-Environment Interaction"]}
    ]
}
```

```python
# Render the Prompt
prompt = template.render(context)
print("Generated Prompt:\n", prompt)
```

```python
client = openai.OpenAI()
```

```python
# OpenAI API Call (GPT-4o-mini)
response = client.chat.completions.create(
    model="gpt-4o-mini",
    messages=[
        {"role": "system", "content": "You are a helpful assistant with mastery in AI."},
        {"role": "user", "content": prompt}
    ],
    logprobs    =True,
    top_logprobs=3
)
```

```python
print(response.choices[0].message.content)
```

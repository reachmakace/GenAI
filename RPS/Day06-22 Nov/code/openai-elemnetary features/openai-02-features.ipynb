````python
import tiktoken
from PIL import Image
````

````python
def estimate_gpt4o_cost(prompt_text: str, image_path: str, estimated_output_tokens: int = 100):
    """
    Estimate cost for using GPT-4o with text + image.
    
    Args:
        prompt_text: Text content (e.g., "Describe this image.")
        image_path: Path to local image file
        estimated_output_tokens: Estimated number of output tokens (default=100)
    
    Returns:
        Dictionary with token/image sizes and estimated cost breakdown.
    """

    # ----- Token cost config -----
    TEXT_INPUT_COST     = 0.005 / 1000  # $ per input token
    TEXT_OUTPUT_COST    = 0.015 / 1000  # $ per output token
    IMAGE_COST_PER_MP   = 0.0075        # $ per megapixel

    # ----- Count input text tokens -----
    #tiktoken.get_encoding
    encoding     = tiktoken.encoding_for_model("gpt-3.5-turbo")
    input_tokens = len(encoding.encode(prompt_text))

    # ----- Image megapixel size -----
    if not os.path.exists(image_path):
        raise FileNotFoundError(f"Image not found: {image_path}")
        
    img           = Image.open(image_path)
    
    width, height = img.size
    megapixels    = (width * height) / 1_000_000

    # ----- Cost Calculations -----
    text_input_cost  = input_tokens            * TEXT_INPUT_COST
    image_input_cost = megapixels              * IMAGE_COST_PER_MP
    text_output_cost = estimated_output_tokens * TEXT_OUTPUT_COST

    total = text_input_cost + image_input_cost + text_output_cost

    return {
        "text_input_tokens": input_tokens,
        "image_dimensions": f"{width}x{height}",
        "image_megapixels": round(megapixels, 2),
        "text_output_tokens_est": estimated_output_tokens,
        "cost_text_input": round(text_input_cost, 6),
        "cost_image_input": round(image_input_cost, 6),
        "cost_text_output": round(text_output_cost, 6),
        "total_cost_usd": round(total, 6)
    }
````

````python
result = estimate_gpt4o_cost(
    prompt_text = "Describe what you see in this picture.",
    image_path  = image_path,
    estimated_output_tokens = 150
)
````

````python
for k, v in result.items():
    print(f"{k}: {v}")
````

**Perform some QnA on image**

````python
from pathlib import Path
````

````python
def encode_image_to_data_uri(image_path: str) -> str:
    image = Path(image_path).read_bytes()
    b64 = base64.b64encode(image).decode("utf-8")
    return f"data:image/png;base64,{b64}"
````

````python
def extract_info_from_image(image_path: str) -> str:
    image_data_uri = encode_image_to_data_uri(image_path)

    system_prompt = """
    You are a clinical document analysis assistant. Extract all readable medical information.
    Provide structured output where possible. Include:
    - Patient Details (Name, Age, Gender, MRN, Date of Visit)
    - Diagnosis
    - Medications
    - Lab/Test Results
    - Doctor Notes
    If any field is missing, respond with null.
    """

    response = client.chat.completions.create(
        model = "gpt-4o",
        messages = [
            {"role": "system", "content": system_prompt},
            {"role": "user", "content": [
                {"type": "image_url", "image_url": {"url": image_data_uri}},
                {"type": "text", "text": "Extract all useful information."}
            ]}
        ]
    )

    output = response.choices[0].message.content.strip()
    return output
````

````python
def ask_question_on_text(text: str, question: str) -> str:
    response = client.chat.completions.create(
        model = "gpt-4o",
        messages = [
            {"role": "system", "content": "You are a helpful clinical assistant. Answer based on provided data."},
            {"role": "user", "content": f"Here is the extracted information:\n\n{text}\n\nQuestion: {question}"}
        ]
    )
    return response.choices[0].message.content.strip()
````

#### Key Limitations of Vision Models in GPT-4o / GPT-4 with Vision

| Category                 | Limitation                                                                 | Practical Implication                                                                         |
| ------------------------ | -------------------------------------------------------------------------- | --------------------------------------------------------------------------------------------- |
| **Medical**           | Not suitable for interpreting **CT, MRI, or specialized images**           | ⚠️ Cannot be used for diagnosis or clinical decision support (e.g., radiology reports)        |
| **Language**          | Non-Latin scripts (e.g., Japanese, Korean, Arabic) may be poorly handled   | OCR accuracy may be low for documents with multilingual or non-Roman scripts                  |
| **Small Text**        | Text that is too small may not be readable                                 | Images with text should be high resolution; enlarge important text where possible             |
| **Rotation**          | Rotated or upside-down images may be misinterpreted                        | Preprocess to **deskew or rotate** images to correct orientation before submission            |
| **Visual Styles**     | Struggles with interpreting **dashed vs. solid lines, color-coded charts** | Style-based distinctions in graphs can be lost; legend comprehension may be poor              |
| **Spatial Reasoning** | Struggles with precise positioning (e.g., chess board, maps)               | Avoid tasks that require object localization or spatial memory                                |
| **Accuracy**          | Can generate plausible but **incorrect or misleading** captions            | Always apply **human verification** or fallback mechanisms, especially in critical workflows  |
| **Image Shape**       | Panoramic, fisheye, or distorted images are poorly handled                 | Standard rectangular images in aspect ratios like 4:3 or 16:9 work best                       |
| **Metadata Ignored**  | File names, EXIF data, and image metadata are **not** used by the model    | You must explicitly **pass any additional metadata** (e.g., timestamps) as text in the prompt |
| **Counting Objects**  | May miscount or give approximate numbers                                   | Don’t rely on it for exact counts of people, items, etc. in an image                          |
| **CAPTCHA Filtering** | Images that resemble CAPTCHA are blocked for safety reasons                | Cannot be used to build CAPTCHA-solving bots — blocked at API level                           |

- Use `gpt-4o-audio-preview` when:
    - You want to mix reasoning and narration in a single request.
    - You’re exploring multimodal outputs in chat flow.
      
- Use `tts-1` or `tts-1-hd` when:
    - You want clean, production-grade audio TTS.
    - You want to directly convert static text (docs, articles, etc.).

When using model="gpt-4o-audio-preview" in a chat completion flow, the model may choose to `summarize`, `paraphrase`, or `rephrase` the input text instead of reading it verbatim — especially for longer inputs. 

That’s because gpt-4o-audio-preview is not just a text-to-speech engine — it's a chat model that also produces audio as a message response. So it tries to "respond" conversationally, not just "read."

| Type              | Price per 1K Tokens |

#### Text-to-Speech (TTS) Use Cases for LLM Applications

**1. Voice Assistants and Chatbots**
- **Use Case**: Give your GPT-powered assistant a human-like voice.
- **Examples**:
  - Customer support bots
  - Voice-enabled smart assistants (e.g., GPT on mobile)
- **Benefit**: Increases accessibility and engagement.

**2. Education & eLearning**
- **Use Case**: Generate spoken explanations or audio lessons from textual content.
- **Examples**:
  - Science lessons or coding tutorials
  - Language learning with pronunciation
- **Benefit**: Supports auditory learners and low-vision students.

**3. Accessibility / Screen Readers**
- **Use Case**: Convert UI content, documents, or summaries into audio.
- **Examples**:
  - Dynamic document narration using GPT + TTS
  - Website accessibility plugins
- **Benefit**: Improves digital inclusion (ADA, WCAG compliance).

**4. Content Creation & Media**
- **Use Case**: Generate narration for videos, podcasts, or audiobooks.
- **Examples**:
  - YouTube voiceovers from GPT scripts

#### Dataset Overview:
- **Name**: MedMCQA
- **Domain**: Medical Question Answering
- **Type**: Multiple-Choice QA (MCQA)
- **Size**: Large-scale dataset with tens of thousands of questions.
- **Purpose**: The dataset is used to train and evaluate models on medical question answering, helping to develop AI systems that can assist with medical exam preparation and knowledge extraction.
  
#### Structure:
- The dataset consists of **multiple-choice questions** with four possible answers.
- For each question, there is a correct answer and additional metadata.

#### Key Features:
1. **Question** (`question`): The medical question in text form, usually related to clinical scenarios, medical knowledge, or diagnostic reasoning.
2. **Answer Choices** (`opa`, `opb`, `opc`, `opd`): Four possible options (A, B, C, and D) from which the model is expected to pick the correct answer.
3. **Correct Answer** (`cop`): The correct option among the four provided choices.
4. **Explanations** (`explanation`): Often, there is an explanation provided to justify the correct answer, offering insight into the reasoning process behind the choice.
5. **Metadata**: Additional fields such as the question type, difficulty, subject area, and any specific medical domain or context.

#### Example Entry:
- **Question**: "Which of the following is the most common cause of fever in children under the age of five?"
- **Option A**: Viral infection
- **Option B**: Bacterial infection
- **Option C**: Fungal infection
- **Option D**: Parasitic infection
- **Correct Answer**: A
- **Explanation**: Viral infections are the most common cause of fever in children under five, followed by bacterial infections.

- **Exam Preparation Tools**: It can be used to create intelligent tutoring systems that assist medical students in preparing for exams.
- **Medical QA Systems**: The dataset is essential for building AI systems capable of understanding medical queries and providing accurate answers, a crucial application in clinical decision support.

#### Creating a Collection in Chroma

- **Purpose**: Chroma uses Collections to store embeddings, documents, and metadata.
- **Method**: Use the `createCollection()` method of the Chroma client.
- **Parameters**:
  - **Collection Name**: Provide a name for the collection.
  - **Optional Embedding Function**: 
    - You can specify an embedding function to generate embeddings from text.
    - If no embedding function is provided, the default embedding function is used.
- **Adding Data to the Collection**:
  - After creating a collection, you need to add your data to it.
  - For each data point, you can add the following:
    - **ID**: Unique identifier for the data point.
    - **Data**: The actual data (e.g., text or document).
    - **Metadata**: Any additional information related to the data point.

#### Distances

**1. Squared L2 Norm**, also known as **L2 Squared** or **Euclidean Distance Squared**, measures the distance between two points in space without the computational cost of a square root.

##### Formula:
For two vectors \( x \) and \( y \), the squared L2 norm is calculated as:
```python
# L2 Squared Formula:
import numpy as np

x = np.array([x1, x2, x3])  # example vector x
y = np.array([y1, y2, y3])  # example vector y

squared_l2 = np.sum((x - y)**2)
print(squared_l2)
```

**Key Points:**
- Avoids square root: Simplifies calculations by omitting the square root operation.
- Distance measure: Represents the squared distance between two points in Euclidean space.
- Common usage: Widely used in machine learning optimization problems, such as linear regression, where exact distances are not required.

**2. inner product**, also known as the **dot product** in Euclidean space, is a measure of similarity between two vectors. It returns a scalar value and is used in many machine learning and linear algebra applications.

##### Formula:
For two vectors \( x \) and \( y \) with \( n \) components, the inner product is calculated as:
$$
x \cdot y = \sum_{i=1}^n x_i y_i
$$

##### Python Example:
```python
import numpy as np

# Define two vectors x and y
x = np.array([x1, x2, x3])
y = np.array([y1, y2, y3])

# Calculate the inner product (dot product)
inner_product = np.dot(x, y)
print(inner_product)

```

**Key Points:**
- Similarity measure: If the inner product is large, the vectors are more aligned; if zero, the vectors are orthogonal.
- Applications: Used in machine learning, neural networks, and geometric computations.


**3. Cosine Similarity** is a metric used to measure the cosine of the angle between two non-zero vectors in an inner product space. It calculates the similarity between two vectors based on their direction, ignoring their magnitude.

For two vectors \( x \) and \( y \), cosine similarity is defined as:
$$
\text{cosine\_similarity}(x, y) = \frac{x \cdot y}{\|x\| \|y\|}
$$

Where:
- $( x \cdot y )$ is the inner product (dot product) of vectors \( x \) and \( y \).
- $( \|x\| ) and ( |y\| )$ are the Euclidean norms (L2 norms) of vectors \( x \) and \( y \).

```python
import numpy as np

# Define two vectors x and y
x = np.array([x1, x2, x3])
y = np.array([y1, y2, y3])

# Calculate cosine similarity
cosine_similarity = np.dot(x, y) / (np.linalg.norm(x) * np.linalg.norm(y))
print(cosine_similarity)

```

**Key Points:**

- Range: The result of cosine similarity lies between -1 and 1.

    - 1 indicates vectors are identical in direction.
    - 0 indicates orthogonality (no similarity).
    - -1 indicates vectors are diametrically opposed.
      
Applications: Commonly used in text analysis, recommendation systems, and other tasks involving high-dimensional data.


**4. Cosine Distance (1 - Cosine Similarity)**

The expression **1 - Cosine Similarity** is used to compute **Cosine Distance**, which measures the dissimilarity between two vectors. While cosine similarity assesses the similarity, cosine distance focuses on how far apart the vectors are.

For two vectors \( x \) and \( y \), the **Cosine Distance** is calculated as:
$$
\text{cosine\_distance}(x, y) = 1 - \frac{x \cdot y}{\|x\| \|y\|}
$$

Where:
- \( x \cdot y \) is the inner product (dot product) of vectors \( x \) and \( y \).
- \( \|x\| \) and \( \|y\| \) are the L2 norms of vectors \( x \) and \( y \).

- **Range**: The cosine distance lies between 0 and 2:
  - **0**: Vectors are identical.
  - **1**: Vectors are orthogonal (no similarity).
  - **2**: Vectors are diametrically opposed.
- **Applications**: Commonly used in clustering, similarity measures, and distance-based algorithms.

```python
import numpy as np

# Define two vectors x and y
x = np.array([x1, x2, x3])
y = np.array([y1, y2, y3])

# Calculate cosine similarity
cosine_similarity = np.dot(x, y) / (np.linalg.norm(x) * np.linalg.norm(y))

# Calculate cosine distance (1 - cosine similarity)
cosine_distance = 1 - cosine_similarity
print(cosine_distance)


#### Cosine Similarity vs Cosine Distance

Below is a comparison between **Cosine Similarity** and **Cosine Distance** in tabular form:

| **Metric**            | **Cosine Similarity**                                   | **Cosine Distance**                                   |
|-----------------------|---------------------------------------------------------|-------------------------------------------------------|
| **Definition**         | Measures the cosine of the angle between two vectors    | Measures how dissimilar two vectors are               |
| **Formula**            | $(\frac{x \cdot y}{\|x\| \|y\|})$                       | $(1 - \frac{x \cdot y}{\|x\| \|y\|})$                 |
| **Range**              | \([-1, 0, 1]\)                                          | \([0, 1, 2]\)                                         |
| **Interpretation**     | 1: Vectors are identical, 0: Vectors are orthogonal, -1: Vectors are opposite | 0: Vectors are identical, 1: Vectors are orthogonal, 2: Vectors are opposite |
| **Use Case**           | Similarity-based algorithms (e.g., document similarity) | Distance-based algorithms (e.g., clustering)          |

##### Key Notes:
- **Cosine Similarity** focuses on how **similar** two vectors are, based on the angle between them.
- **Cosine Distance** focuses on how **dissimilar** two vectors are, derived by subtracting cosine similarity from 1.

`Cosine Similarity vs L2 Distance:`

- **Cosine Similarity**:
  - Compares the angle between two vectors to measure similarity, ignoring magnitude differences.
  - Suitable for text or document embeddings where direction matters more than vector length.

- **L2 Distance (Euclidean)** (default):
  - Measures the straight-line distance between vectors, and is sensitive to both direction and magnitude.
  - Often used in geometric or spatial data comparisons.

- `Alternatives to "hnsw" in ChromaDB`

In ChromaDB, **"hnsw"** refers to the **Hierarchical Navigable Small World** algorithm, which is commonly used for fast approximate nearest neighbor searches. However, ChromaDB supports other algorithms and methods for vector search and storage.

Here are some alternatives:

1. **"flat"**:
   - The **flat index** performs an exact nearest neighbor search.
   - This is slower compared to "hnsw" but guarantees exact results.
   - Suitable when speed is less of a concern, and accuracy is paramount.

2. **"ivf"** (Inverted File Index):
   - Combines clustering and approximate search.
   - Vectors are grouped into clusters, and searching happens only in the nearest cluster.
   - It’s faster than "flat" for large datasets but provides approximate results.
  
3. **"pq"** (Product Quantization):
   - Compresses vectors into smaller codes, reducing memory usage.
   - Useful for scenarios where memory efficiency is a priority over search accuracy.
   - Offers an approximate nearest neighbor search like "hnsw."

**Summary of Alternatives:**
- **hnsw**: Fast approximate nearest neighbor search (default).
- **flat**: Exact nearest neighbor search with slower performance.
- **ivf**: Clustered approximate search for large datasets.
- **pq**: Compressed vectors for memory efficiency and approximate searches.

#### Explanation of `collection_cosine.add()`

- **`collection_cosine.add()`**: This function adds data to the ChromaDB collection named `collection_cosine`. The function expects **IDs**, **documents**, and **metadata** for each entry.

Here's a detailed breakdown of each parameter used:

1. **`ids=[str(i) for i in range(0, 100)]`**:
   - Generates a list of 100 IDs as strings from 0 to 99.
   - Each entry in the collection requires a unique ID, which is used to reference or retrieve the document later.
   - The list comprehension `str(i) for i in range(0, 100)` creates these unique string IDs (`'0'`, `'1'`, ..., `'99'`).

2. **`documents=dataset_cleaned["exp"][:100]`**:
   - **`dataset_cleaned["exp"]`** refers to the cleaned dataset containing explanations in the `"exp"` column.
   - The slice `[:100]` selects the first 100 explanation entries.
   - These explanations are stored as **documents** in the collection. The document field usually contains the main content or text you want to create embeddings for.

3. **`metadatas=[{"type": "exp"} for _ in range(0, 100)]`**:
   - A list of metadata for each document.
   - Here, it creates 100 identical metadata dictionaries, each containing `{"type": "exp"}`, signifying that the document type is an explanation (`"exp"`).
   - Metadata provides additional information about the document, which can be useful for categorization, filtering, or retrieval.

#### Summary:
- The `collection_cosine.add()` function stores **100 entries** in the `collection_cosine`.
- Each entry consists of:
  - A unique string **ID** (from 0 to 99),
  - An explanation **document** (from the cleaned dataset),
  - **Metadata** indicating the document type.
  
This prepares the collection to perform vector searches (such as cosine similarity searches) using the added explanations.

#### Introduction and Description 
---

**Recall@k** is a fundamental evaluation metric for ranking and recommendation systems that measures how well a system captures relevant items within the top-k results. 

Unlike Precision@k which focuses on the quality of recommendations, Recall@k emphasizes **coverage** - how many of the total relevant items were successfully found in the top-k predictions.

**Key Concepts:**

- **Coverage-based metric**: Measures what fraction of relevant items were captured
- **Completeness-focused**: Evaluates how well the system avoids missing relevant items
- **Denominator varies**: Unlike Precision@k, denominator is total relevant items (not k)

**How it differs from Precision@k:**

| Metric | Question Asked | Focus | Denominator |
|--------|----------------|-------|-------------|
| **Precision@k** | "Of the k items I showed, how many were relevant?" | Quality | k (fixed) |
| **Recall@k** | "Of all relevant items, how many did I find in top-k?" | Coverage | Total relevant items |

This makes Recall@k particularly valuable when:
- **Missing relevant items** is costly (medical diagnosis, security screening)
- **Comprehensive coverage** is more important than precision
- You want to ensure **no important items are overlooked**
- **Discovery and exploration** are key user goals

#### Mathematical Definition 
---

The formula for Recall@k is:

$$\text{Recall@k} = \frac{\text{Number of relevant items in top-k predictions}}{\text{Total number of relevant items}}$$

More formally:

$$\text{Recall@k} = \frac{|\text{Relevant} \cap \text{Top-k}|}{|\text{Relevant}|}$$

Where:
- $|\text{Relevant} \cap \text{Top-k}|$ = number of relevant items found in top-k predictions
- $|\text{Relevant}|$ = total number of relevant items that exist

**Key Properties:**
- **Range**: [0, 1] where 1 means all relevant items were found
- **Higher is better**: More relevant items captured means better coverage
- **Variable denominator**: Depends on total relevant items (user-dependent)
- **Monotonic**: Recall@k can only increase or stay same as k increases
- **Upper bound**: Cannot exceed 1.0, achieved when all relevant items are in top-k

#### Practical Illustration 
---

Let's work through a concrete example to understand how Recall@k works.

**Scenario: Medical Diagnosis System**

A patient presents symptoms, and there are **8 total relevant conditions** that could explain their symptoms. Our diagnostic system ranks possible conditions and we want to evaluate Recall@5.

**All Relevant Conditions (Ground Truth):**
1. "Viral Pneumonia" ✅
2. "Bacterial Pneumonia" ✅  
3. "Bronchitis" ✅
4. "Asthma Exacerbation" ✅
5. "Heart Failure" ✅
6. "Pulmonary Embolism" ✅
7. "COVID-19" ✅
8. "Lung Cancer" ✅

**System's Top-5 Predictions:**
1. "Viral Pneumonia" ✅ (Found relevant condition)
2. "Common Cold" ❌ (Missed - not in relevant list)
3. "Bronchitis" ✅ (Found relevant condition)
4. "Seasonal Allergies" ❌ (Missed - not in relevant list)
5. "Heart Failure" ✅ (Found relevant condition)

**Recall@5 Calculation:**
- **Relevant items found in top-5**: 3 (Viral Pneumonia, Bronchitis, Heart Failure)
- **Total relevant items**: 8
- **Recall@5** = 3/8 = 0.375 or **37.5%**

**Interpretation:**
The system captured only 37.5% of all relevant conditions in its top-5 recommendations. This means **5 important conditions were missed** (Bacterial Pneumonia, Asthma, Pulmonary Embolism, COVID-19, Lung Cancer), which could be dangerous in a medical context.

**Why this matters:**
- Missing "Pulmonary Embolism" could be life-threatening
- Missing "COVID-19" has public health implications
- High recall is crucial when missing relevant items has serious consequences

#### Python Implementation
---

Let's implement Recall@k from scratch and demonstrate it with examples.

````python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from typing import List, Union

# Set up plotting style
plt.style.use('default')
sns.set_palette("husl")
np.random.seed(42)
````

````python
def recall_at_k(actual: List[Union[int, str]], 
                predicted: List[Union[int, str]], 
                k: int) -> float:
    """
    Calculate Recall@k for a single query/user.
    
    Args:
        actual: List of actual relevant items
        predicted: List of predicted items in ranked order
        k: Number of top predictions to consider
    
    Returns:
        Recall@k score (between 0 and 1)
    """
    if len(actual) == 0:
        return 0.0
    
    if k <= 0:
        return 0.0
    
    # Get top-k predictions
    top_k_predictions = set(predicted[:k])
    actual_set = set(actual)
    
    # Count how many actual relevant items are in top-k predictions
    relevant_found = len(actual_set.intersection(top_k_predictions))
    
    return relevant_found / len(actual)


def precision_at_k(actual: List[Union[int, str]], 
                   predicted: List[Union[int, str]], 
                   k: int) -> float:
    """
    Calculate Precision@k for comparison with Recall@k.
    """
    if k <= 0:
        return 0.0
    
    top_k_predictions = predicted[:k]
    relevant_count = sum(1 for item in top_k_predictions if item in actual)
    
    return relevant_count / k


def f1_at_k(actual: List[Union[int, str]], 
            predicted: List[Union[int, str]], 
            k: int) -> float:
    """
    Calculate F1@k - harmonic mean of Precision@k and Recall@k.
    """
    precision = precision_at_k(actual, predicted, k)
    recall    = recall_at_k(actual, predicted, k)
    
    if precision + recall == 0:
        return 0.0
    
    return 2 * (precision * recall) / (precision + recall)
````

````python
#### Practical Illustration
# All relevant conditions for this patient
actual_conditions = [
    "Viral Pneumonia", 
    "Bacterial Pneumonia", 
    "Bronchitis", 
    "Asthma Exacerbation", 
    "Heart Failure", 
    "Pulmonary Embolism", 
    "COVID-19", 
    "Lung Cancer"
]

# System's ranked predictions
predicted_conditions = [
    "Viral Pneumonia",      # ✅ Relevant
    "Common Cold",          # ❌ Not relevant
    "Bronchitis",           # ✅ Relevant  
    "Seasonal Allergies",   # ❌ Not relevant
    "Heart Failure",        # ✅ Relevant
    "Bacterial Pneumonia",  # ✅ Relevant
    "Gastritis",            # ❌ Not relevant
    "COVID-19",             # ✅ Relevant
    "Asthma Exacerbation",  # ✅ Relevant
    "Lung Cancer"           # ✅ Relevant
]
````

````python
print(f"Total relevant conditions: {len(actual_conditions)}")
print(f"System predictions available: {len(predicted_conditions)}")
````

````python
# Calculate metrics for different k values
k_values = [1, 3, 5, 7, 10]
print("k  | Recall@k | Precision@k | F1@k   | Interpretation")
print("-" * 65)

for k in k_values:
    recall    = recall_at_k(actual_conditions, predicted_conditions, k)
    precision = precision_at_k(actual_conditions, predicted_conditions, k)
    f1        = f1_at_k(actual_conditions, predicted_conditions, k)
    
    if k == 1:
        interp = "Good start"
    elif k == 3:
        interp = "Low coverage"
    elif k == 5:
        interp = "Still missing many"
    elif k == 7:
        interp = "Better coverage"
    else:
        interp = "High coverage"
    
    print(f"{k:2d} | {recall:8.3f} | {precision:11.3f} | {f1:6.3f} | {interp}")
````

#### Recall@k vs Precision@k
---

Understanding the relationship and trade-offs between these complementary metrics:

**🔄 The Precision-Recall Trade-off:**

**When to Prioritize Recall@k:**
- **High cost of missing items**: Medical diagnosis, security screening
- **Discovery applications**: Research, exploration, brainstorming
- **Comprehensive search**: Legal discovery, academic literature review
- **Safety-critical systems**: Risk assessment, threat detection

**When to Prioritize Precision@k:**
- **Limited user attention**: Mobile apps, quick decisions
- **High cost of false positives**: Spam detection, fraud alerts
- **Quality over quantity**: Premium recommendations, expert systems
- **Resource constraints**: Limited processing time, bandwidth

**When to Balance Both (F1@k):**
- **General recommendation systems**: E-commerce, content platforms
- **Search engines**: Web search, document retrieval
- **Business applications**: CRM, sales leads, marketing
- **Educational systems**: Learning resources, course recommendations

**Framework for Setting k in Recall@k:**

**Step 1: Define Coverage Requirements**
```
Questions to ask:
• What percentage of relevant items must be captured?
• What's the cost of missing important items?
• Are there regulatory or safety requirements?
• How comprehensive does the search need to be?
```

**Step 2: Estimate Total Relevant Items**
```
Consider:
• Average number of relevant items per query/user
• Distribution of relevant items (some users have many, others few)
• Domain-specific patterns and seasonality
• System's ability to rank relevant items highly
```

**Step 3: Calculate Required k**
```
Formula approach:
• If you want 80% recall and expect 20 relevant items on average
• And your system ranks 60% of relevant items in top positions
• You need k ≈ (0.8 × 20) / 0.6 = 27
```

**Step 4: Balance with Practical Constraints**
```
Considerations:
• Computational cost of processing large k
• User interface and cognitive load
• Business model and monetization
• Storage and bandwidth limitations
```

#### Ground Truth Acquisition: The Challenge of Knowing "All Relevant Items"
---

**The Ground Truth Problem** is the biggest practical challenge with Recall@k. 

Unlike Precision@k where you only need to evaluate the items you showed, Recall@k requires knowing **all possible relevant items** - which is often impossible or expensive to obtain.

**Methods for Ground Truth Acquisition:**

**1. Expert Annotation (Gold Standard)**
- **Method**: Domain experts manually identify all relevant items
- **Quality**: Highest accuracy and reliability
- **Cost**: Very expensive, time-consuming
- **Scalability**: Poor - doesn't scale to large datasets
- **Use cases**: Medical diagnosis, legal documents, academic research

**2. User Behavior Data (Implicit Feedback)**
- **Method**: Infer relevance from user actions (clicks, purchases, time spent)
- **Quality**: Noisy but realistic
- **Cost**: Low - collected automatically
- **Scalability**: Excellent - scales with user base
- **Challenges**: Selection bias, missing implicit signals

**3. Crowdsourcing Approaches**
- **Method**: Multiple non-expert annotators label relevance
- **Quality**: Good with proper quality control
- **Cost**: Moderate - cheaper than experts
- **Scalability**: Good - can parallelize work
- **Challenges**: Quality control, annotator agreement


#### Advantages and Limitations {#advantages-and-limitations}
---

**✅ Advantages of Recall@k:**

1. **Coverage-Focused**
   - Measures how well system captures all relevant items
   - Critical for applications where missing items is costly

2. **User-Centric for Discovery**
   - Reflects users' desire to see all available options
   - Important for exploration and research scenarios

3. **Safety and Compliance**
   - Essential for medical, legal, and safety-critical applications
   - Helps ensure regulatory compliance and risk management

4. **Monotonic Property**
   - Recall@k can only increase or stay same as k increases
   - Predictable behavior makes it easier to reason about

5. **Complementary to Precision**
   - Provides different perspective on system performance
   - Combined with Precision@k gives complete picture

**❌ Limitations of Recall@k:**

1. **Requires Known Ground Truth**
   - Need complete list of relevant items for each query/user
   - Often difficult or expensive to obtain in practice

2. **Variable Denominator**
   - Total relevant items varies across users/queries
   - Makes comparison across different scenarios challenging

3. **Ignores Ranking Quality**
   - Doesn't consider position of relevant items within top-k
   - Relevant item at position 1 same as position k

4. **No Quality Assessment**
   - Focuses on quantity, not quality of recommendations
   - May encourage systems to include marginal items

5. **Can Be Misleading**
   - High recall with very large k may not be practically useful
   - Doesn't account for user attention and cognitive limits

**🔄 Alternative and Complementary Metrics:**

- **Precision@k**: Quality of top-k recommendations
- **F1@k**: Harmonic mean balancing precision and recall
- **NDCG@k**: Considers both relevance and ranking position
- **MAP@k**: Mean Average Precision across multiple k values
- **Coverage**: Fraction of item catalog recommended across all users
- **Diversity@k**: Variety of different types of items in recommendations

#### What is Mean Reciprocal Rank (MRR)?

**Intuition:**
Mean Reciprocal Rank (MRR) is a metric used to evaluate how well a system ranks the correct answer for a set of queries. It is especially useful in search engines, question answering, and recommendation systems where you want the correct result to appear as high as possible in the list of returned items.

MRR focuses on the position of the first correct answer in the ranked list. The higher the correct answer appears, the better the score. If the correct answer is at the top (rank 1), the reciprocal rank is 1. If it is at position 2, the reciprocal rank is 0.5, and so on.

**Example 1:**
Suppose a search engine returns the following ranked results for a query, and the correct answer is at position 3:

1. Result A
2. Result B
3. **Result C** (correct)
4. Result D

The reciprocal rank for this query is $1/3 = 0.333$.

**Example 2:**
For another query, the correct answer is at the top:

1. **Result X** (correct)
2. Result Y
3. Result Z

The reciprocal rank is $1/1 = 1.0$.

**MRR Calculation:**
If you have multiple queries, MRR is the average of the reciprocal ranks for all queries. This rewards systems that consistently rank the correct answer near the top.

**Why use MRR?**
- It is simple and interpretable.
- It directly measures how quickly a user finds the correct answer.
- It is ideal when only one correct answer matters per query.

**Mathematical Formula**

$$
\text{MRR} = \frac{1}{|Q|} \sum_{i=1}^{|Q|} \frac{1}{\text{rank}_i}
$$

Where $|Q|$ is the number of queries, and $\text{rank}_i$ is the rank position of the first correct answer for query $i$.

**Use Cases**
- Search engines (ranking relevant documents)
- Question answering (ranking correct answers)
- Recommendation systems (ranking relevant items)

**When to Apply**
- When the position of the first correct result is important
- Each query has a single correct answer
- Evaluating ranking models or retrieval systems

**Pros and Cons**
**Pros:**
- Simple to compute and interpret
- Focuses on the first relevant result
- Useful for single-answer retrieval tasks

**Cons:**
- Ignores all but the first relevant result
- Not suitable for multi-answer queries
- Sensitive to ranking errors at the top

#### Python Implementation of MRR

```python
def mean_reciprocal_rank(list_of_ranked_lists, ground_truths):
    """
    list_of_ranked_lists: List of lists, each containing ranked items for a query
    ground_truths: List of correct answers for each query
    Returns: MRR score
    """
    reciprocal_ranks = []
    for ranked_list, truth in zip(list_of_ranked_lists, ground_truths):
        try:
            rank = ranked_list.index(truth) + 1
            reciprocal_ranks.append(1.0 / rank)
        except ValueError:
            reciprocal_ranks.append(0.0)
    return sum(reciprocal_ranks) / len(reciprocal_ranks)
```

```python
# Example usage
queries = [
    ['cat', 'dog', 'mouse'],        # ranked list for query 1
    ['apple', 'banana', 'orange'],  # ranked list for query 2
]

answers = ['dog', 'orange']  # correct answers

mrr_score = mean_reciprocal_rank(queries, answers)

print(f"MRR: {mrr_score:.3f}")
```

**Best Practices**
- Use MRR for tasks where the first correct result is most important.
- For multi-answer queries, consider metrics like Mean Average Precision (MAP).
- Combine MRR with other metrics for a comprehensive evaluation.

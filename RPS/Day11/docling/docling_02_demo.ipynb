```python
import os

try:
    import docling
except ImportError:
    docling = None
    
from PyPDF2 import PdfReader
```

#### Load a Sample PDF Document

We'll load a sample PDF file and extract its text content.

```python
# Load a sample PDF and extract text
pdf_path = r"../data/rules_of_ml.pdf"  # Change to any sample PDF in your data folder
reader   = PdfReader(pdf_path)

pdf_text = "\n".join(page.extract_text() or "" for page in reader.pages)
print(f"Extracted {len(pdf_text)} characters from PDF.")
```

#### Chunk the PDF Content

Now, we'll split the extracted PDF text into manageable chunks. If Docling is not available, we'll use a simple sentence splitter as a fallback.

```python
# Chunk the PDF text
if docling and hasattr(docling, 'chunk_text'):
    chunks = docling.chunk_text(pdf_text)
else:
    # Fallback: simple sentence splitter (by period)
    import re
    chunks = re.split(r'(?<=[.!?])\s+', pdf_text)
    
print(f"Total chunks: {len(chunks)}")
```

#### Display the Extracted Chunks

Let's print the first few chunks to inspect the results.

```python
# Display the first 5 chunks
for i, chunk in enumerate(chunks[:5]):
    print(f"Chunk {i+1}:")
    print(chunk)
    print("-"*40)
```

#### Simple conversion

- What this example does

- Converts a single source (URL or local file path) to a unified Docling document and prints Markdown to stdout.

```python
# takes time to load (first time)
from docling.document_converter import DocumentConverter
```

```python
# takes time to load
# Change this to a local path or another URL if desired.
# Note: using the default URL requires network access; if offline, provide a
# local file path (e.g., Path("/path/to/file.pdf")).
source = "https://arxiv.org/pdf/2408.09869"

converter = DocumentConverter()
result    = converter.convert(source)

# Print Markdown to stdout.
print(result.document.export_to_markdown())
```

```python
# Chunk the Markdown output from the medical guidelines PDF
markdown_text = result.document.export_to_markdown()

# Example: Split by headings (e.g., '## ')
import re
chunks = re.split(r'(^## .*$)', markdown_text, flags=re.MULTILINE)

# Combine heading and content for each chunk
chunked_sections = []
for i in range(1, len(chunks), 2):
    heading = chunks[i].strip()
    content = chunks[i+1].strip() if i+1 < len(chunks) else ''
    chunked_sections.append(f"{heading}\n{content}")

print(f"Total chunks: {len(chunked_sections)}")
for i, chunk in enumerate(chunked_sections[:5]):
    print(f"Chunk {i+1} (first 300 chars):\n", chunk[:300], "\n---\n")
```
